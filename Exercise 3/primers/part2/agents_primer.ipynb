````python
import os
import io
import pandas as pd

def gather_summary(df):
    # 2. Capture df.info() output
    buffer = io.StringIO()
    df.info(buf=buffer)
    info_text = buffer.getvalue()
    print(info_text)

    # 3. Calculate missing value stats
    missing_stats = (df.isna().sum() / len(df) * 100).sort_values(ascending=False)
    missing_summary = "\n".join([f"{col}: {val:.2f}%" for col, val in missing_stats.items()])
    print(missing_summary)

    # 4. Get column data types
    column_types = "\n".join([f"{col}: {dtype}" for col, dtype in df.dtypes.items()])
    print(column_types)

    # 5. Get unique value counts
    unique_counts = df.nunique()  # Will no longer fail on unhashable dict
    unique_counts_summary = "\n".join([f"{col}: {count}" for col, count in unique_counts.items()])
    print(unique_counts_summary)

    summary_text = f"""
    Dataset Name: {dataset_name}
    ----------------------------
    Shape: {df.shape[0]} rows x {df.shape[1]} columns
    
    Column Data Types:
    {column_types}
    
    Missing Value Percentage:
    {missing_summary}
    
    Unique Value Counts:
    {unique_counts_summary}
    
    Data (first {n_sample} rows):
    {df.head(n_sample).to_string()}
    
    Data Description:
    {df.describe().to_string()}
    
    Data Info:
    {info_text}
    """
    print(summary_text)

    return summary_text
````

````python
dataset_name = "weatherAUS.csv"
````

````python
n_sample=30
summary_text = ""
if os.path.exists(dataset_name):
    df = pd.read_csv(dataset_name)
    df = df.convert_dtypes()
    summary_text = gather_summary(df)
````

````python
def clean_data(data_raw):
    import pandas as pd
    import numpy as np
    from sklearn.impute import SimpleImputer
    
    # Create a copy to avoid modifying the original data
    data = data_raw.copy()
    
    print(f"Initial shape: {data.shape}")
    
    # Step 1: Remove columns with excessive missing values (>40%)
    # Drop Sunshine (48.01%), Evaporation (43.17%), Cloud3pm (40.81%)
    columns_to_drop = ['Sunshine', 'Evaporation', 'Cloud3pm']
    data = data.drop(columns=columns_to_drop)
    print(f"After removing columns with >40% missing: {data.shape}")
    
    # Step 2: Convert Date column to datetime format
    data['Date'] = pd.to_datetime(data['Date'])
    print("Date column converted to datetime")
    
    # Step 3: Remove duplicate rows
    initial_rows = len(data)
    data = data.drop_duplicates()
    print(f"Duplicates removed: {initial_rows - len(data)} rows")
    
    # Step 4: Handle categorical columns with missing values (impute with mode)
    categorical_columns = ['WindGustDir', 'WindDir9am', 'WindDir3pm', 'RainToday', 'RainTomorrow']
    for col in categorical_columns:
        if data[col].isnull().sum() > 0:
            mode_value = data[col].mode()[0]
            data[col] = data[col].fillna(mode_value)
    print(f"Categorical columns imputed with mode")
    
    # Step 5: Handle numeric columns with missing values (impute with mean)
    numeric_columns = ['Cloud9am', 'Pressure9am', 'Pressure3pm', 'WindGustSpeed', 
                      'Humidity3pm', 'Temp3pm', 'Rainfall', 'WindSpeed3pm', 
                      'Humidity9am', 'WindSpeed9am', 'Temp9am', 'MinTemp', 'MaxTemp']
    
    # Use SimpleImputer for numeric columns
    imputer = SimpleImputer(strategy='mean')
    for col in numeric_columns:
        if col in data.columns and data[col].isnull().sum() > 0:
            # fit_transform returns 2D array, use ravel() to convert to 1D
            data[col] = imputer.fit_transform(data[[col]]).ravel()
    print(f"Numeric columns imputed with mean")
    
    # Step 6: Remove rows with remaining missing values
    rows_before = len(data)
    data = data.dropna()
    print(f"Rows with remaining missing values removed: {rows_before - len(data)} rows")
    
    # Step 7: Detect and remove outliers using IQR method (3x IQR)
    numeric_cols_for_outliers = ['MinTemp', 'MaxTemp', 'Rainfall', 'WindGustSpeed', 
                                  'WindSpeed9am', 'WindSpeed3pm', 'Humidity9am', 
                                  'Humidity3pm', 'Pressure9am', 'Pressure3pm', 
                                  'Temp9am', 'Temp3pm']
    
    rows_before = len(data)
    for col in numeric_cols_for_outliers:
        if col in data.columns:
            Q1 = data[col].quantile(0.25)
            Q3 = data[col].quantile(0.75)
            IQR = Q3 - Q1
            lower_bound = Q1 - 3 * IQR
            upper_bound = Q3 + 3 * IQR
            data = data[(data[col] >= lower_bound) & (data[col] <= upper_bound)]
    print(f"Outliers removed (3x IQR method): {rows_before - len(data)} rows")
    
    # Step 8: Validate logical consistency (Custom Step)
    # Ensure MinTemp <= MaxTemp
    rows_before = len(data)
    data = data[data['MinTemp'] <= data['MaxTemp']]
    print(f"Invalid temperature ranges removed: {rows_before - len(data)} rows")
    
    # Ensure Temp9am and Temp3pm fall within MinTemp and MaxTemp range
    rows_before = len(data)
    data = data[(data['Temp9am'] >= data['MinTemp']) & (data['Temp9am'] <= data['MaxTemp'])]
    data = data[(data['Temp3pm'] >= data['MinTemp']) & (data['Temp3pm'] <= data['MaxTemp'])]
    print(f"Invalid temp readings removed: {rows_before - len(data)} rows")
    
    # Step 9: Validate categorical values (Custom Step)
    # Valid wind directions
    valid_wind_directions = ['N', 'NNE', 'NE', 'ENE', 'E', 'ESE', 'SE', 'SSE', 
                             'S', 'SSW', 'SW', 'WSW', 'W', 'WNW', 'NW', 'NNW']
    
    rows_before = len(data)
    data = data[data['WindGustDir'].isin(valid_wind_directions)]
    data = data[data['WindDir9am'].isin(valid_wind_directions)]
    data = data[data['WindDir3pm'].isin(valid_wind_directions)]
    print(f"Invalid wind directions removed: {rows_before - len(data)} rows")
    
    # Valid rain values
    rows_before = len(data)
    data = data[data['RainToday'].isin(['Yes', 'No'])]
    data = data[data['RainTomorrow'].isin(['Yes', 'No'])]
    print(f"Invalid rain indicator values removed: {rows_before - len(data)} rows")
    
    # Step 10: Check for temporal consistency (Custom Step)
    # Sort by Location and Date to check temporal order
    data = data.sort_values(['Location', 'Date']).reset_index(drop=True)
    print("Data sorted by Location and Date for temporal consistency")
    
    # Step 11: Final validation
    print(f"\nFinal shape: {data.shape}")
    print(f"Missing values remaining: {data.isnull().sum().sum()}")
    print(f"Data types correct: Date is {data['Date'].dtype}")
    
    return data
````
